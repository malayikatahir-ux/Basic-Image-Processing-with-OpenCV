```python
# Load image
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread('images.jpg')
plt.imshow(img)
plt.show()
```

```python
# Shape
print("Shape of Image",img.shape)
```

```python
# Resizing
re = cv2.resize(img,(150,150))
plt.imshow(re)
plt.show()
```

```python
# Croping
crop = img[80:180, 45:100]
plt.imshow(crop)
plt.axis("off")
plt.show()
```

```python
# Rotating 90
ro = cv2.rotate(img,cv2.ROTATE_90_CLOCKWISE)
plt.imshow(ro)
plt.axis("off")
plt.show()
```

```python
# Rotating Anticlockwise
ro = cv2.rotate(img,cv2.ROTATE_90_COUNTERCLOCKWISE)
plt.imshow(ro)
plt.axis("off")
plt.show()
```

```python
# Converting to GrayScale
gray = cv2.cvtColor(img,cv2.COLOR_RGB2GRAY)
plt.imshow(gray)
plt.axis("off")
plt.show()
```

```python
gray = cv2.cvtColor(img,cv2.COLOR_RGB2GRAY)
plt.imshow(gray,cmap="gray")
plt.axis("off")
plt.show()
```

```python
# converting to RGB
rgb = cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
plt.imshow(rgb)
plt.axis("off")
plt.show()
```

```python
# converting to HSV
hs = cv2.cvtColor(img,cv2.COLOR_RGB2HSV)
plt.imshow(hs)
plt.axis("off")
plt.show()
```

```python
# saving
cv2.imwrite("Original image.jpg",img)
cv2.imwrite("Resized image.jpg",re)
cv2.imwrite("Cropped image.jpg",crop)
cv2.imwrite("Rotated image.jpg",ro)
cv2.imwrite("GrayScale image.jpg",gray)
cv2.imwrite("RGB image.jpg",rgb)
cv2.imwrite("HSV image.jpg",hs)
```
